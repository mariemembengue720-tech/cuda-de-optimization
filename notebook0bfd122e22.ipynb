```python
import subprocess
import re
import numpy as np
import pandas as pd

#  Paramètres complets du sujet
func_names = {
    1: "Shifted Rastrigin",
    2: "Shifted Rosenbrock",
    3: "Shifted Griewank",
    4: "Shifted Sphere"
}

dims = [10, 50, 100]
pops = [50, 100, 500]
modes = ["cpu", "gpu", "gpu-fused", "gpu-optimized"]
seeds = list(range(1, 11))  # 🌱 10 exécutions indépendantes

summary_data = []

print(" Lancement du benchmark complet (36 configurations x 4 modes x 10 runs)...")

for obj_func, func_name in func_names.items():
    for dim in dims:
        for pop in pops:
            print(f" Test : {func_name} | Dim: {dim} | Pop: {pop}")

            cpu_times = {}  # Stocke les temps CPU par graine pour calculer le speedup exact

            for mode in modes:
                times = []
                fitnesses = []

                for seed in seeds:
                    cmd = f"./de_app {mode} {dim} {pop} {obj_func} {seed}"
                    res = subprocess.run(cmd, shell=True, capture_output=True, text=True)

                    time_match = re.search(r"Time:\s*([\d\.]+)\s*ms", res.stdout)
                    fit_match = re.search(r"Best Fitness:\s*([-\d\.]+)", res.stdout)

                    if time_match and fit_match:
                        t = float(time_match.group(1))
                        f = float(fit_match.group(1))
                        times.append(t)
                        fitnesses.append(f)

                        if mode == "cpu":
                            cpu_times[seed] = t

                if times:
                    mean_time = np.mean(times)
                    std_time = np.std(times)
                    mean_fit = np.mean(fitnesses)
                    std_fit = np.std(fitnesses)

                    #  Calcul du speedup moyen graine par graine
                    if mode == "cpu":
                        mean_speedup = 1.0
                    else:
                        speedups = [cpu_times[s] / t for s, t in zip(seeds, times) if s in cpu_times]
                        mean_speedup = np.mean(speedups)

                    summary_data.append({
                        "Fonction ": func_name,
                        "Dimension ": dim,
                        "Population ": pop,
                        "Mode ": mode,
                        "Temps (ms) ": f"{mean_time:.2f} ± {std_time:.2f}",
                        "Speedup Moyen ": f"{mean_speedup:.2f}x",
                        "Best Fitness ": f"{mean_fit:.2e} ± {std_fit:.2e}"
                    })

#  Affichage et Sauvegarde des Résultats
df_summary = pd.DataFrame(summary_data)

# Export CSV
df_summary.to_csv("benchmark_de_results.csv", index=False)

# Export LaTeX pour le rapport
with open("benchmark_de_results.tex", "w") as f:
    f.write(df_summary.to_latex(index=False))

print("\n✅ Benchmark terminé ! Fichiers 'benchmark_de_results.csv' et 'benchmark_de_results.tex' générés avec succès.")
```
